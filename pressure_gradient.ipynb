```python
import numpy as np

# Define constants and grid
rho = 1.225  # Air density in kg/m^3 (at sea level)
dx = 100.0  # Distance step in km

# Pressure values along an x-axis (in hecterPascals)
# High pressure on the left, low pressure on the right
pressure = np.array([1015, 1014, 1012, 1009, 1005])

# Calculate the pressure gradient using finite differences (dp/dx)
dp_dx = np.gradient(pressure, dx)

# Calculate Pressure Gradient Force per unit mass (m/s^2)
pgf_x = -(1.0 / rho) * dp_dx

print("Pressure Gradient (hPa/m):", dp_dx)
print("Pressure Gradient Force (m/s^2):", pgf_x)
```
